# Czyszczenie – wszystkie pliki

## 1. Konfiguracja

```python
import sys
sys.path.append("..")

import pandas as pd
from tqdm import tqdm

from src.cleaning import clean_file
from src.config import PROCESSED_DIR, STUDY_FILES

FILES = STUDY_FILES
CLEAN_DIR = PROCESSED_DIR / "clean"
WEEKLY_DIR = PROCESSED_DIR / "weekly_counts"
FUNNEL_DIR = PROCESSED_DIR / "funnel"
for folder in (CLEAN_DIR, WEEKLY_DIR, FUNNEL_DIR):
    folder.mkdir(parents=True, exist_ok=True)
```

## 2. Czyszczenie plików

```python
summary = []
pending = [(s, c) for s, c in FILES if not (CLEAN_DIR / f"r_{s}_comments_{c}.parquet").exists()]
print(f"Skipping {len(FILES) - len(pending)} existing files, {len(pending)} to process")

with tqdm(pending, desc="Pliki", unit="plik") as bar:
    for subreddit, cycle in bar:
        bar.set_postfix_str(f"{subreddit} {cycle}")
        clean, weekly, funnel = clean_file(subreddit, cycle)
        name = f"r_{subreddit}_comments_{cycle}"
        clean.to_parquet(CLEAN_DIR / f"{name}.parquet", index=False)
        weekly.to_csv(WEEKLY_DIR / f"{name}.csv", index=False)
        funnel.to_csv(FUNNEL_DIR / f"{name}.csv", index=False)
        summary.append({"subreddit": subreddit, "cycle": cycle, "n_in_window": len(clean), "n_keep": int(clean["keep"].sum())})
        del clean, weekly, funnel
```

## 3. Podsumowanie

```python
pd.DataFrame(summary)
```
