# 04 Rysunki – polaryzacja (H4a, H4b)

## 1. Konfiguracja

```python
import sys
sys.path.append("..")

import matplotlib.pyplot as plt
import pandas as pd

from src.config import APA_STYLE, CYCLES, FIGURES_DIR, POLITICAL_SUBREDDITS, RESULTS_DIR, SUBREDDIT_STYLES

FIGURES_DIR.mkdir(parents=True, exist_ok=True)
plt.rcParams.update(APA_STYLE)
```

## 2. WPA w fazach szczytowej i przejściowej (H4a)

```python
wpa = pd.read_csv(RESULTS_DIR / "h4a_wpa.csv")
wpa = wpa[(wpa["scope"] == "all") & (wpa["variant"] == "wpa")]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(6.3, 3.3), sharey=True)
for ax, (phase, title) in zip(axes, (("szczytowa", "Faza szczytowa"), ("przejsciowa", "Faza przejściowa"))):
    for subreddit in POLITICAL_SUBREDDITS:
        rows = wpa[(wpa["subreddit"] == subreddit) & (wpa["phase"] == phase)].set_index("cycle").reindex(CYCLES)
        color, linestyle = SUBREDDIT_STYLES[subreddit]
        ax.plot(CYCLES, rows["wpa"], color=color, linestyle=linestyle, marker="o", markersize=4, label=subreddit)
    ax.set_title(title, fontsize=10)
    ax.set_xticks(CYCLES)
    ax.set_xlabel("Cykl")
axes[0].set_ylabel("WPA")
axes[0].legend(frameon=False, fontsize=8)
fig.tight_layout()
fig.savefig(FIGURES_DIR / "wpa.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 3. Trafność klasyfikacji obozów (H4b)

```python
summary = pd.read_csv(RESULTS_DIR / "h4b_summary.csv").set_index("cycle")
```

```python
fig, ax = plt.subplots(figsize=(6.3, 3.3))
ax.errorbar(range(len(CYCLES)), summary.loc[CYCLES, "mean"],
            yerr=[summary.loc[CYCLES, "mean"] - summary.loc[CYCLES, "p2_5"], summary.loc[CYCLES, "p97_5"] - summary.loc[CYCLES, "mean"]],
            color="black", marker="o", markersize=4, capsize=3, linestyle="none", label="Trafność (średnia i 2,5.–97,5. percentyl)")
ax.plot(range(len(CYCLES)), summary.loc[CYCLES, "acc_dem"], color="black", linestyle="--", marker="s", markersize=3, label="Demokraci")
ax.plot(range(len(CYCLES)), summary.loc[CYCLES, "acc_rep"], color="0.45", linestyle=":", marker="^", markersize=3, label="Republikanie")
ax.axhline(0.5, color="black", linewidth=0.6, linestyle=(0, (1, 4)))
ax.set_xticks(range(len(CYCLES)), CYCLES)
ax.set_xlabel("Cykl")
ax.set_ylabel("Trafność klasyfikacji")
ax.legend(frameon=False, fontsize=8)
fig.tight_layout()
fig.savefig(FIGURES_DIR / "h4b_accuracy.png", dpi=300, bbox_inches="tight")
plt.show()
```
